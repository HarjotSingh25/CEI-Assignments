```python
import re
import json
import nltk
from collections import Counter

nltk.download('punkt')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))
```

```python
def keyword_agent(text):

    words = word_tokenize(text.lower())

    words = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]

    freq = Counter(words)

    keywords = [word for word, count in freq.most_common(5)]

    return keywords
```

```python
def summary_agent(text):

    sentences = sent_tokenize(text)

    if len(sentences) <= 2:
        return text

    return " ".join(sentences[:2])
```

```python
def question_agent(text):

    keywords = keyword_agent(text)

    questions = []

    for word in keywords:
        questions.append(f"What is {word}?")

    return questions
```

```python
positive_words = {
    "good","excellent","great","happy","improve","success",
    "benefit","positive","growth","powerful","efficient","best"
}

negative_words = {
    "bad","poor","failure","negative","loss","problem",
    "issue","difficult","error","risk","weak"
}


def sentiment_agent(text):

    words = word_tokenize(text.lower())

    positive = sum(word in positive_words for word in words)
    negative = sum(word in negative_words for word in words)

    if positive > negative:
        return "Positive"

    elif negative > positive:
        return "Negative"

    else:
        return "Neutral"
```

```python
def planner_agent(user_query):

    query = user_query.lower()

    if "summary" in query:
        return "summary"

    elif "keyword" in query:
        return "keywords"

    elif "question" in query:
        return "questions"

    elif "sentiment" in query:
        return "sentiment"

    else:
        return "analysis"
```

```python
def router_agent(intent, document):

    result = {}

    print("\nPlanner Agent")
    print("Intent Detected:", intent)

    print("\nRouter Agent")

    if intent == "summary":

        print("Running Summary Agent")

        result["Summary"] = summary_agent(document)

    elif intent == "keywords":

        print("Running Keyword Agent")

        result["Keywords"] = keyword_agent(document)

    elif intent == "questions":

        print("Running Question Agent")

        result["Questions"] = question_agent(document)

    elif intent == "sentiment":

        print("Running Sentiment Agent")

        result["Sentiment"] = sentiment_agent(document)

    else:

        print("Running Full Analysis Pipeline")

        result["Keywords"] = keyword_agent(document)

        result["Summary"] = summary_agent(document)

        result["Questions"] = question_agent(document)

        result["Sentiment"] = sentiment_agent(document)

    return result
```

```python
def report_agent(intent, output):

    report = {

        "Intent": intent,

        "Output": output

    }

    print("\nReport Agent")

    print(json.dumps(report, indent=4))

    return report
```

```python
def smart_document_assistant(query, document):

    print("="*60)

    print("SMART DOCUMENT ASSISTANT")

    print("="*60)

    print("\nInput Agent")

    print("Document Received")

    intent = planner_agent(query)

    result = router_agent(intent, document)

    final_report = report_agent(intent, result)

    return final_report
```

```python
document = """
Artificial Intelligence is transforming healthcare. AI helps doctors diagnose diseases quickly and accurately. Machine learning models analyze medical records to improve patient care. AI also reduces hospital costs and improves treatment planning. These technologies are making healthcare more efficient and reliable.
"""
```

```python
query = input("Enter your request: ")

smart_document_assistant(query, document)
```
